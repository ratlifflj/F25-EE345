### Part III: K-Means algorithm

(a) Given a list of vectors data and a list of k centroids, write a function `group_assignment` that assigns each vector to the centroid closest to it.

```python
"""
The function below returns

grouping_vec_c: The grouping of each data point in data to single representative centroid.

This is a list of integers of the same length as data. Each element of the list contains the
index of the chosen centroid for the corresponding data point.
"""

def group_assignment(data,centroids):
    #TODO: PART B
    grouping_vec_c = np.zeros(len(data))
    
    #For each data point
    for i in range(len(data)):
        
        #Find distance to all centroids
        dist = np.zeros(len(centroids))
        
        for j in range(len(centroids)):
            dist[j] = np.linalg.norm(data[i] - centroids[j])
        
        #Compute centroid assignment
        min_dist = min(dist)
        for j in range(len(centroids)):
            if min_dist == dist[j]:
                grouping_vec_c[i] = j+1
    
    return grouping_vec_c
```

(c) Given a list of vectors data, their grouping and the list a list of k centroids, write a function `update_centroid` that updates the centroids based on the grouping.

```python
"""
The function below returns

new_centroids: The new centroid for the grouping.

This is a list of coordinates of the same length as centroids. 
"""

def update_centroid(data, grouping, centroids):
    new_centroids = []
    #TODO: PART C
    for i in range(len(centroids)):
        cent = np.zeros(len(data[0]))
        count = 0
        for j in range(len(data)):
            if grouping[j] == (i+1):
                cent = cent+data[j]
                count += 1
        group_average = cent/count
        new_centroids.append(group_average)
    return new_centroids

# The objective function for clustering written for you
def clustering_objective(data, grouping, centroids):
    J_obj = 0
    for i in range(len(data)):
        for j in range(len(centroids)):
            if grouping[i] == (j+1):
                J_obj += np.linalg.norm(data[i] - centroids[j])**2
    
    
    J_obj = J_obj/len(data)
    return J_obj
```

(d) Using the provided functions `update_centroid` and `clustering_objective`, implement the k-means algorithm which terminates when distance the chosen centroids at successive iterations is less than $10^{−6}$.

```python
"""
Runs K-Means algorithm until termination. Return the following

new_centroids: the centroids chosen at the end of the algorithm
grouping: the grouping of each of the data points
J_obj_vector: a record of the clustering objective at each iteration
iteration: total number of iterations run by the algorithm
"""

def Kmeans_alg(data, centroids):
    #TODO: PART D
    iteration = 0
    J_obj_vector = []
    Stop = False
    
    
    while Stop == False:
        #ipdb.set_trace()
        grouping = group_assignment(data, centroids)
        new_centroids = update_centroid(data, grouping, centroids)
        J_obj = clustering_objective(data, grouping, new_centroids)
        J_obj_vector.append(J_obj)
        iteration += 1
        
        #Termination condition - Error is small enough
        if np.linalg.norm(np.array(new_centroids) - np.array(centroids)) < 1e-8:
            Stop = True
        else:
            centroids = new_centroids

    return new_centroids, grouping, J_obj_vector, iteration
```

Make sure your functions are working, the following cell will run the Kmeans algorithm for k=3 and you should be able to see 3 cluster 

```python
num_points = len(X)
random_idxs = np.random.choice(num_points, 3, replace=False)
centroids = X[random_idxs]
new_centroids, grouping, J_obj_vector, iteration = Kmeans_alg(X, centroids)
plt.scatter(X[:,0], X[:,1],c = grouping)
```

```python
K_values = [1,2,3,4,5,6]
# TODO: Part E: Run the algorithm for different k values
""" Hint: It might help to store the final objective function and 
     the grouping corresponding to each k value in arrays/lists since you 
     will use them to make your plots
"""
# This will hold the grouping array for each k
Groups = []
# This will hold the value of the objective function for each k
Objs =[]
num_points = len(X)
for k in K_values:
    Groups_candidates =[]
    Obj_candidates =[]
    for iter in range(10):
        random_idxs = np.random.choice(num_points, k,replace=False)
        centroids = X[random_idxs]
        new_centroids, grouping, J_obj_vector, iteration = Kmeans_alg(X, centroids)
        Groups_candidates.append(grouping)
        Obj_candidates.append(J_obj_vector[-1])
    min_i = np.argmin(Obj_candidates)
    Groups.append(Groups_candidates[min_i])
    Objs.append(Obj_candidates[min_i])
```
